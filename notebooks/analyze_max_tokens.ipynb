```python
import os
import pandas as pd
```

```python
#base_path = "../outputs/gemma3_1b_cl_paper/"
#base_path = "../outputs/qwen35_08b_cl_paper/"
base_path = "../outputs/llama32_1b_cl_paper/"
#base_path = "../outputs/qwen35_08b_paper_321/"
#base_path = "../outputs/gemma3_1b_cl_paper_321/"
#base_path = "../outputs/llama32_1b_cl_paper_321/"
test_predictions_file = "test_predictions.jsonl"

for dir in os.listdir(base_path):
    
    if not os.path.isdir(os.path.join(base_path, dir)):
        continue

    if "eval" not in dir:
        continue

    predictions_path = os.path.join(base_path, dir, test_predictions_file)

    df = pd.read_json(predictions_path, lines=True)
    count_hit_max_tokens = len(df[df["hit_max_tokens"] == True])

    if count_hit_max_tokens > 0:
        print(dir)
        print(f"Count: {len(df)}, Hit max tokens: {count_hit_max_tokens}")
        print()
```
