```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("meta-llama/Llama-3.2-1B-Instruct")

messages = [{"role": "user", "content": "Test prompt"}]

rendered = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True,
)

print(rendered)
```

```python
rendered_a = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True,
    date_string="26 Jul 2024",
)

rendered_b = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True,
    date_string="31 Aug 2026",
)

ids_a = tokenizer(rendered_a, add_special_tokens=False).input_ids
ids_b = tokenizer(rendered_b, add_special_tokens=False).input_ids

print(rendered_a == rendered_b)  # False
print(ids_a == ids_b)            # False
```
